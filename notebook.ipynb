Every year, American high school students take SATs, which are standardized tests intended to measure literacy, numeracy, and writing skills. There are three sections - reading, math, and writing, each with a **maximum score of 800 points**. These tests are extremely important for students and colleges, as they play a pivotal role in the admissions process.

Analyzing the performance of schools is important for a variety of stakeholders, including policy and education professionals, researchers, government, and even parents considering which school their children should attend. 

You have been provided with a dataset called `schools.csv`, which is previewed below.

You have been tasked with answering three key questions about New York City (NYC) public school SAT performance.

```python
# Re-run this cell 
import pandas as pd

# Read in the data
schools = pd.read_csv("schools.csv")

# Preview the data
schools.head()
```

```python
#get null columns
schools.isna().any()
```

```python
#get the best math results are at least 80% of the *maximum possible score of 800* for math
#the best math results that's more than 640
best_math=schools[schools["average_math"]>=640]
#create dateframe for best math schools and sorted by "average_math" in descending order
best_math_schools=best_math[["school_name","average_math"]].sort_values("average_math",ascending=False)
best_math_schools.head()
```

```python
#sc=schools.groupby("school_name")["average_math",'average_writing','average_reading'].sum()
#sorted by average_math
#sc=schools.sort_values("average_math",ascending=False)
```

```python
#add new column total_SAT 
schools['total_SAT']=schools["average_math"]+schools['average_writing']+schools['average_reading']
schools.head()
#get the top 10 
top=schools.sort_values("total_SAT",ascending=False)
top_10=top.iloc[:10,]
#create dataframe with 2 columns (school name and total sat )
top_10_schools=top_10[["school_name","total_SAT"]]
top_10_schools.head()
```

```python
# Group by borough and calculate the required aggregations
borough_stats = schools.groupby('borough')['total_SAT'].agg(['count', 'mean', 'std']).round(2)

# Find the borough with the largest standard deviation
largest_std_dev_borough = borough_stats.nlargest(1, 'std')

# Rename columns for clarity and to match the request
largest_std_dev = largest_std_dev_borough.rename(columns={
    'count': 'num_schools',
    'mean': 'average_SAT',
    'std': 'std_SAT'
})

# rest borough as a column
largest_std_dev.reset_index()

# Reorder the columns to match the requested order
largest_std_dev = largest_std_dev[[ 'num_schools', 'average_SAT', 'std_SAT']]
largest_std_dev.head()
```
